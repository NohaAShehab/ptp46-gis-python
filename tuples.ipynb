<h1> tuples </h1>
- tuples are immutable datatype --> once created cannot be changed
- can different values from different datatypes

```python
t = ()
```

```python
myt = tuple()
```

```python
myt = ("iti",23,23.243,False, ["python", "arcGIS", "iti"],
       'iti', True, None, "iti", ('iti', 'GIS'))
```

```python
# access element using using index
print(myt[3])
```

```python
myt[3:]  # return new tuple
```

```python
print(myt[::-1]) # new tuple
```

```python
print(len(myt))
```

```python
print(myt.count("iti"))
```

```python
print(myt[4])
```

```python
myt[4][2]
```

```python
# update value at existing index ?
myt[3]="update value"
print(myt)
```

```python
print(myt.index("iti")) # get index of the first occurrence of the element
```

```python
# iterable

for item in myt:
    print(f"item = {item}")
```

```python
for index, item in enumerate(myt):
    print(f"index = {index} , item = {item}")
```

<h1 style='color:yellow; text-align:center;'>Encapsulation </h1>

---
### 📚 **Encapsulation & access in Python**
Python has **no** access modifiers like `public`/`private`/`protected`. We use **naming conventions**:
- <span style="color:#27ae60;">**public**</span> → normal names <span style="color:#95a5a6;">[a-z, ...]</span>
- <span style="color:#f39c12;">**protected**</span> → single underscore <span style="color:#f39c12;">_name</span>
- <span style="color:#e74c3c;">**private**</span> → double underscore <span style="color:#e74c3c;">__name</span> (name mangling)

```python
"""
    encapsulation ::: access modifiers ?
    --> public
    --> protected
    --> private

    Python says that NO ACCESS MODIFIERS IN PYTHON
    # use name convention ?
    public ---> [a-z: ]
    protected ---> [_:]
    private ---> [__:]

"""
```

---
### 👤 **`Employee` with three visibility levels**
| Attribute | Convention | Visibility |
|-----------|------------|------------|
| <span style="color:#27ae60;">name</span> | no underscore | **public** |
| <span style="color:#f39c12;">_email</span> | single `_` | **protected** |
| <span style="color:#e74c3c;">__salary</span> | double `__` | **private** (name-mangled) |
Creating an instance: <span style="color:#3498db;">Employee("noha", 'noha@gmail.com', 50000)</span>

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.__salary = salary


emp = Employee("noha", 'noha@gmail.com',50000)
```

---
### 🟢 **Public attribute — full access**
<span style="color:#27ae60;">name</span> is **public**: you can <span style="color:#3498db;">print(emp.name)</span> and <span style="color:#9b59b6;">emp.name = 'updated'</span>. No restrictions.

```python
print(emp.name) # public
emp.name = 'updated'
print(emp.name)
```

- Ethically don't this ? don't access protected members using object

---
### 🟡 **Accessing a protected member**
<span style="color:#f39c12;">_email</span> is **protected** (single leading underscore). You can still do <span style="color:#3498db;">print(emp._email)</span>; linters may warn that you’re touching a “protected” member.

```python
# protected
print(emp._email) # Access to a protected member _email of a class
```

---
### ✏️ **Modifying a protected member**
We *can* assign <span style="color:#e67e22;">emp._email = 'updated@gmail.com'</span>. Protected is only a **convention** — Python does not enforce it. Use it to signal “internal use.”

```python
emp._email = 'updated@gmail.com'
```

- access private members

---
### 🚫 **Private attribute — direct access fails**
<span style="color:#e74c3c;">emp.__salary</span> raises <span style="color:#c0392b;">AttributeError</span> because Python renames it to <span style="color:#3498db;">_Employee__salary</span> (name mangling). You cannot access it by the name you wrote.

```python
print(emp.__salary)
```

---
### ❌ **Missing attribute**
<span style="color:#e74c3c;">emp.city</span> was never set — <span style="color:#c0392b;">AttributeError</span>. Unlike private name mangling, this is simply “no such attribute.”

```python
print(emp.city)
```

---
### 🔍 **Inspecting `__dict__`**
<span style="color:#3498db;">__dict__</span> shows the real attribute names. Notice <span style="color:#e74c3c;">__salary</span> becomes <span style="color:#27ae60;">_Employee__salary</span> — **name mangling** prevents accidental override in subclasses.

```python
print(emp.__dict__)
```

### 🔓 **Bypassing name mangling (not recommended)**
Private attribute is stored as <span style="color:#e74c3c;">_Employee__salary</span>. You *can* access it as <span style="color:#95a5a6;">emp._Employee__salary</span>, but this breaks encapsulation — avoid in real code.

<h1 style='color:pink; text-align:center;'>Introducing Getters and setters </h1>

---
### 📤 **Getter method**
A method that **returns** the private <span style="color:#3498db;">__salary</span>. Call it with <span style="color:#27ae60;">emp.salary()</span> — no direct access to the attribute, so you control how it’s read.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.__salary = salary


    def salary(self):
        return self.__salary

emp = Employee("noha", 'noha@gmail.com', 50000)
print(emp.salary())
```

---
### 🛡️ **Getter + setter with validation**
<span style="color:#27ae60;">salary()</span> returns the value; <span style="color:#3498db;">set_salary(salary)</span> checks <span style="color:#9b59b6;">isinstance(salary, int) and salary > 0</span> before updating. Invalid input raises <span style="color:#e74c3c;">TypeError</span>.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.__salary = salary


    def salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be a float")


emp = Employee("noha", 'noha@gmail.com', 50000)
print(emp.salary())

emp.set_salary(45645)

emp2= Employee("sadds", "aadad", "dasdads")
```

---
### ⚠️ **Validation in `__init__`**
Invalid <span style="color:#e74c3c;">salary</span> (e.g. <span style="color:#e67e22;">"dasdads"</span>) is rejected when creating the object. <span style="color:#c0392b;">TypeError: Salary must be a float</span> — constructor enforces the same rules as the setter.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be a float")


    def salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be an integer")



emp2= Employee("sadds", "aadad", "dasdads")
```

---
### 🎯 **Public vs private in practice**
<span style="color:#27ae60;">emp.name = 'updated'</span> — direct access to **public** member. <span style="color:#9b59b6;">emp.salary()</span> and <span style="color:#9b59b6;">emp.set_salary(3248)</span> — **private** member via getter/setter. Validation only on the private path.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be a float")


    def salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be an integer")



emp2= Employee("sadds", "aadad", 234)
# setting object property == instance variables
emp.name = 'updated' # deals with public member
print(emp.salary())
emp.set_salary(3248)  # deals with private member
```

<h1 style='color:green;'> Property decorator </h1>

---
### 🌟 **The `@property` decorator**
Turns a method into an **attribute**: read <span style="color:#27ae60;">emp2.salary</span> instead of <span style="color:#95a5a6;">emp2.salary()</span>. The private <span style="color:#3498db;">__salary</span> is still hidden; the property is the public face.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be a float")

    @property  # this function represent the property
    # you can directly call it instanceName.salary
    def salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be an integer")



emp2= Employee("sadds", "aadad", 234)
print(emp2.salary)
print(emp2.salary())
```

---
### 📦 **Property = attribute, not method**
With <span style="color:#9b59b6;">@property</span>, use <span style="color:#27ae60;">emp2.salary</span> (no parentheses). Calling <span style="color:#e67e22;">emp2.salary()</span> fails — it’s an `int`, not callable. <span style="color:#3498db;">__dict__</span> still shows <span style="color:#95a5a6;">_Employee__salary</span>.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be a float")

    @property  # this function represent the property
    # you can directly call it instanceName.salary
    def salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be an integer")



emp2= Employee("sadds", "aadad", 234)
print(emp2.salary)
print(emp.__dict__)
```

---
### ✏️ **Property setter**
<span style="color:#e74c3c;">@salary.setter</span> lets you write <span style="color:#3498db;">emp2.salary = 5000</span> instead of <span style="color:#95a5a6;">emp2.set_salary(5000)</span>. Same validation, cleaner syntax.

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be a float")

    @property  # this function represent the property
    # you can directly call it instanceName.salary
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be an integer")



emp2= Employee("sadds", "aadad", 234)
print(emp2.salary)
print(emp2.__dict__)
emp2.salary = 5000
print(emp2.salary)
```

---
### ⚙️ **Property + assignment in `__init__`**
Using `self.salary = salary` in `__init__` calls the **setter**, so validation runs at construction too. The setter stores into `__salary`.
<span style="color:#1abc9c;">self.salary = salary</span> → triggers setter

```python
class Employee:
    def __init__(self, name, email, salary):
        self.name = name
        self._email = email
        self.salary = salary


    @property  # this function represent the property
    # you can directly call it instanceName.salary
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("Salary must be an integer")



emp2= Employee("sadds", "aadad", 234)
print(emp2.salary)
print(emp2.__dict__)
emp2.salary = 5000
print(emp2.salary)
```

---
### 📐 **Computed / derived property**
`age` is not stored — it’s **computed** from `birthyear`. Access with `s.age` (no parentheses). No setter ⇒ read-only.
<span style="color:#9b59b6;">@property</span> → getter only | <span style="color:#3498db;">2026 - birthyear</span>

```python
class Student:
    def __init__(self, name, email, birthyear):
        self.name = name
        self.email = email
        self.birthyear = birthyear

    @property
    def age(self):
        return 2026 - self.birthyear


s = Student("sadds", "aadad", 2000)
print(s.age)
print(s.__dict__)
```

---
### 🔒 **Read-only property — no setter**
Assigning to `s.age` fails because the property has no setter. The computed value is **read-only**.
<span style="color:#e67e22;">AttributeError</span>: property has no setter

```python
s.age = 29
```

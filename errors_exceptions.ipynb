```python
import math
try:
    num1 = int(input("Please enter a number: "))
    num2 = int(input("Please enter a number: "))
    res = num1/ num2
    print(f" res = {res}")
except ValueError as e :
    print(f"--- error happened --- {e}")
    print("num1, num2 must be integers")
    res = None
except ZeroDivisionError as ze :
    print("-- You cannot divide by zero --")
    res = math.inf
except Exception as e :
    print(f"--- error happened --- {e}")
    exit()
```

# Introducing else block

```python
try:
    num1 = int(input("Please enter a number: "))
    num2 = int(input("Please enter a number: "))
    res = num1/ num2

except Exception as e :
    print(f"--- error happened --- {e}")
else:
    print("--- process completed successfully ---- ")
    print(f'res=  {res}')
```

```python
def ask_for_int(message = "please enter an integer "):
    try:
        num = int(input(message))
    except Exception as e:
        return None
    else:
        return num
    finally:
        print("------ thank you for using our app ---------")
    print("************************************")

print(ask_for_int("please enter an integer "))
```

```python
def sum_num(num1, num2):
    if not( isinstance(num1, int) and isinstance(num2, int)):
        raise TypeError("num1, num2 must be integers")
    else:
        print(f"num1= {num1}, num2= {num2}")
        print("--- performming the summation ---")
        res = num1 + num2
        print(f"res= {res}")
        return  res
```

### 👤 **Simple class and instance**
Define <span style="color:#27ae60;">Employee</span> with <span style="color:#3498db;">__init__(self, name, email, salary)</span> and a method <span style="color:#9b59b6;">display_employee()</span>. Create an instance with <span style="color:#e74c3c;">emp = Employee(...)</span> and inspect attributes with <span style="color:#1abc9c;">emp.__dict__</span>.

```python
class Employee:
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary


    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

emp = Employee("Ahmed", "ahmed@gmail.com", 2000)
emp.display_employee()

print(emp.__dict__)
```

<h1 style='color:orange; text-align:center'> Count number of objects created from the class </h1>

---
### ❌ **Wrong way: instance variable as counter**
<span style="color:#e74c3c;">self.count = 0</span> and <span style="color:#e67e22;">self.count += 1</span> create a **new** counter per instance (each starts at 1). To count **all** objects, we need a **class** variable, not an instance variable.

```python
class Employee:
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary
        self.count = 0
        self.count +=1


    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

emp2 = Employee("Ali", "Ali@gmail.com", 3000)

emp3 = Employee("Adam", "Adam@gmail.com", 4455)
```

---
### 📈 **Class variable — shared counter**
<span style="color:#27ae60;">count = 0</span> is a **class variable** (one per class). <span style="color:#3498db;">Employee.count += 1</span> in <span style="color:#9b59b6;">__init__</span> increments it for every new instance. <span style="color:#e74c3c;">emp.__dict__</span> has no <span style="color:#1abc9c;">count</span> — it lives on the class.

```python
# define property represent class itself not the objects created

class Employee:
    count = 0 # class variable represent class property
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary
        Employee.count += 1

    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

print(Employee.count)
emp2 = Employee("Ali", "Ali@gmail.com", 3000)
print(Employee.count)
emp = Employee("Ahmed", "ahmed@gmail.com", 2000)
print(Employee.count)
emp3 = Employee("Adam", "Adam@gmail.com", 4455)
print(Employee.count)
print(emp.__dict__)
```

---
### 🔗 **`self.__class__.count` and instance `id`**
<span style="color:#3498db;">self.__class__.count</span> is the same as <span style="color:#27ae60;">Employee.count</span> (dynamic class reference). <span style="color:#9b59b6;">self.id = Employee.count</span> gives each instance a numeric ID. <span style="color:#e74c3c;">__dict__</span> shows <span style="color:#1abc9c;">id: 2</span> for the second employee.

```python
# define property represent class itself not the objects created

class Employee:
    count = 0 # class variable represent class property
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary
        print(Employee, self.__class__)
        self.__class__.count += 1
        self.id = Employee.count

    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

print(Employee.count)
emp2 = Employee("Ali", "Ali@gmail.com", 3000)
print(Employee.count)
emp = Employee("Ahmed", "ahmed@gmail.com", 2000)
print(Employee.count)
emp3 = Employee("Adam", "Adam@gmail.com", 4455)
print(Employee.count)
print(emp.__dict__)
```

<h1 style="color:purple; text-align:center">Class methods </h1>

### 🎯 **`@classmethod` — first param is `cls`**
<span style="color:#9b59b6;">@classmethod</span> makes the first parameter the **class** (<span style="color:#3498db;">cls</span>), not the instance. <span style="color:#27ae60;">get_number_of_employees(cls)</span> returns <span style="color:#e74c3c;">cls.count</span>. Call with <span style="color:#1abc9c;">Employee3.get_number_of_employees()</span> — no instance needed.

```python
# define property represent class itself not the objects created
# Python decorators ? --> used before function >> modify its default behaviour
class Employee3:
    count = 0 # class variable represent class property
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.count += 1
        self.id = Employee.count

    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

    # you are special function that represent class behaviour or class action

    @classmethod
    def get_number_of_employees(cls): # Usually first parameter of such methods is named 'cls'
        print(cls)
        return cls.count

    def get_count(self):
        return Employee.count

# class methods --> call it using className
# print(Employee3.get_number_of_employees())

print(Employee3.count)

```

---
### 📊 **Class method — same idea, different class name**
Same pattern as before: <span style="color:#3498db;">@classmethod</span> <span style="color:#27ae60;">get_number_of_employees(cls)</span> receives the class. <span style="color:#e74c3c;">Employee3.count</span> is the class variable. (Note: <span style="color:#95a5a6;">self.id = Employee.count</span> is a bug if Employee is another class.)

```python
# define property represent class itself not the objects created
# Python decorators ? --> used before function >> modify its default behaviour
class Employee3:
    count = 0 # class variable represent class property
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.count += 1
        self.id = Employee.count

    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

    # you are special function that represent class behaviour or class action

    @classmethod
    def get_number_of_employees(cls): # Usually first parameter of such methods is named 'cls'
        print(cls)
        return cls.count

    def get_count(self):
        return Employee.count

# class methods --> call it using className
# print(Employee3.get_number_of_employees())

print(Employee3.count)

```

### 🏭 **`@classmethod` as factory**
<span style="color:#9b59b6;">create_default_object(cls)</span> returns <span style="color:#27ae60;">cls("test", "new", 213244)</span> — an instance of the class. Call as <span style="color:#3498db;">Employee4.create_default_object()</span>. <span style="color:#e74c3c;">cls</span> is the class (e.g. Employee4), so subclasses get their own type.

```python
# define property represent class itself not the objects created
# Python decorators ? --> used before function >> modify its default behaviour
class Employee4:
    count = 0 # class variable represent class property
    def __init__(self, name, email,salary):
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.count += 1
        self.id = Employee4.count

    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

    # you are special function that represent class behaviour or class action

    @classmethod
    def create_default_object(cls):
        # cls === Employee4
        # return Employee4("test", "new", 213244)
        return cls("test", "new", 213244)

# class methods --> call it using className
# print(Employee3.get_number_of_employees())

emp = Employee4.create_default_object()
emp2 = Employee4("noha", "noha@gmail.com", 32424)

```

### 🏭 **Class method as factory (with default args)**
<span style="color:#3498db;">__init__(self, name="", email="", salary=6565)</span> has **defaults**, so <span style="color:#27ae60;">create_default_object()</span> can do <span style="color:#9b59b6;">return cls("test", "new", 213244)</span> and also support <span style="color:#e74c3c;">Employee4("noha", "noha@gmail.com", 32424)</span>.

```python
# define property represent class itself not the objects created
# Python decorators ? --> used before function >> modify its default behaviour
class Employee4:
    count = 0 # class variable represent class property
    def __init__(self, name="", email="",salary=6565):
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.count += 1
        self.id = Employee4.count

    def display_employee(self):
        print(f"Employee Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

    # you are special function that represent class behaviour or class action

    @classmethod
    def create_default_object(cls):
        # cls === Employee4
        # return Employee4("test", "new", 213244)
        return cls("test", "new", 213244)

# class methods --> call it using className
# print(Employee3.get_number_of_employees())

emp = Employee4.create_default_object()
emp2 = Employee4("noha", "noha@gmail.com", 32424)

```

<h1 style="color:orange; text-align:center">Static methods </h1>


---
### 🧮 **Helper logic and static methods**
<span style="color:#e74c3c;">cal_net_salary(salary)</span> is a standalone function that computes net salary. We use it with <span style="color:#3498db;">mm.salary</span>. Later we’ll move it inside the class as a <span style="color:#9b59b6;">@staticmethod</span>.

```python
class Manager:
    count = 0 # class variable represent class property
    def __init__(self, name="", email="",salary=6565):
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.count += 1
        self.id = Manager.count

    def display_employee(self):
        print(f"Manager Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

    @classmethod
    def create_default_object(cls):
        return cls("test", "new", 213244)

mm = Manager("test", "test@gmail.com", 56564)

def cal_net_salary(salary):
    if isinstance(salary, int):
        return  salary* .8
    else:
        raise Exception("Salary must be an integer")

cal_net_salary(10000)
```

---
### 📤 **Using a standalone function**
<span style="color:#3498db;">cal_net_salary(mm.salary)</span> — a regular function that takes a number. It works but isn’t attached to the class; next we’ll make it a <span style="color:#27ae60;">@staticmethod</span> for better organization.

```python
cal_net_salary(mm.salary)
```

---
### 🔧 **`@staticmethod` — helper on the class**
<span style="color:#9b59b6;">cal_net_salary(salary)</span> is a **static** method: no <span style="color:#e74c3c;">self</span>, no <span style="color:#3498db;">cls</span>. Call as <span style="color:#27ae60;">Manager.cal_net_salary(mm.salary)</span>. It’s a utility tied to the class but not to instance or class state.

```python
class Manager:
    count = 0 # class variable represent class property
    def __init__(self, name="", email="",salary=6565):
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.count += 1
        self.id = Manager.count

    def display_employee(self):
        print(f"Manager Name: {self.name}, Email: {self.email}, Salary: {self.salary}")

    @classmethod
    def create_default_object(cls):
        return cls("test", "new", 213244)

    # helper function not related to the class or to the object
    @staticmethod
    def cal_net_salary(salary):
        """
        this a helper method to calculate the net salary for any given salary
        :param salary:
        :return: net salary
        """
        if isinstance(salary, int):
            return  salary* .8
        else:
            raise Exception("Salary must be an integer")

mm = Manager("test", "test@gmail.com", 56564)
Manager.cal_net_salary(mm.salary)
```

---
### 📞 **Calling static method with any value**
<span style="color:#27ae60;">Manager.cal_net_salary(sal)</span> — no instance needed. Pass any number; the method is a <span style="color:#3498db;">staticmethod</span> that belongs to the class namespace and doesn’t use <span style="color:#e74c3c;">self</span> or <span style="color:#9b59b6;">cls</span>.

```python
sal = 324352
print(Manager.cal_net_salary(sal))
```

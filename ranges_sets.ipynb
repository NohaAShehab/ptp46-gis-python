# Ranges

```python
# ranges

# I need create list of numbers from 1 to 10

nums = []
counter  = 1

while counter <= 10:
    nums.append(counter)
    counter += 1

print(nums)
```

```python
nums = []
counter  = 1

while counter <= 10:
    nums.append(counter)
    counter += 2

print(nums)
```

```python
# sequence based datatype range from --> to , step

# range -> for integers

myr = range(10)
print(myr)
```

```python
# iterable object
for i in myr:
    print(i)
```

```python
# cast range to a list
myl = list(myr)
print(myl)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(100, 10, 2):
    print(i)
```

```python
for i in range(100, 10, -2):
    print(i)
```

```python
# generate list of numbers in one line

ll = [num for num in range(10)]
print(ll )
```

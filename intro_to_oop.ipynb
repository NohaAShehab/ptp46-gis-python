```python
class Student:
    """define constructor method"""
    def __init__(self): # you need new address for this object {self}
        "name, email, salary --> instance variables --> their value depends on the caller instance"
        self.name = 'Name'
        self.email = 'Email'
        self.salary = 1000

s = Student()
print(s)

s2 = Student()
print(s2)
```

```python

class Student:

    def __init__(self, name, email='test', salary=100):
        self.name = name
        self.email = email
        self.salary = salary


    def display_student(self):
        print(f"student name: {self.name}, email: {self.email}, salary: {self.salary}")


s = Student("new", "new@gmail.com", 2000)
s.display_student()

s2 = Student("Ahmed")
s2.display_student()
```

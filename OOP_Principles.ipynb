<h1 style='color:red;'> Inheritance </h1>

```python
class Parent:
    pass


class Child(Parent):
    pass

c = Child()
print(isinstance(c, Parent))
```

```python
class Parent:
    def __init__(self, name):
        self.name = name
        print("parent is created")

    def display(self):
        print(f"Parent name is {self.name}")

class Child(Parent):
    pass

c = Child()
print(isinstance(c, Parent))
```

```python
c = Child("ahmed")
print(isinstance(c, Parent))
c.display() # call parent
```

<h1 style='color:yellow;'> Multiple Inheritance </h1>

```python
class A:
    pass

class B:
    pass


class C(A, B):
    pass

cc = C()
print(isinstance(cc, A))
print(isinstance(cc, B))
```

```python
class A:
    def __init__(self):
        print("A is created")

class B:
    pass


class C(A, B):
    pass

cc = C()
```

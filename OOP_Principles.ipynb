<h1 style='color:red;'> Inheritance </h1>

---
### 🧬 **Basic inheritance**
<span style="color:#27ae60;">Child(Parent)</span> means <span style="color:#3498db;">Child</span> inherits from <span style="color:#e74c3c;">Parent</span>. <span style="color:#9b59b6;">isinstance(c, Parent)</span> is <span style="color:#1abc9c;">True</span> — a <span style="color:#27ae60;">Child</span> instance is also a <span style="color:#e74c3c;">Parent</span>.

```python
class Parent:
    pass


class Child(Parent):
    pass

c = Child()
print(isinstance(c, Parent))
```

---
### ⚠️ **Parent requires an argument**
<span style="color:#3498db;">Parent.__init__(self, name)</span> needs <span style="color:#e74c3c;">name</span>. <span style="color:#e67e22;">Child()</span> with no args fails: <span style="color:#c0392b;">TypeError: Parent.__init__() missing 1 required positional argument: 'name'</span>. Child inherits the same constructor.

```python
class Parent:
    def __init__(self, name):
        self.name = name
        print("parent is created")

    def display(self):
        print(f"Parent name is {self.name}")

class Child(Parent):
    pass

c = Child()
print(isinstance(c, Parent))
```

---
### 📞 **Child inherits parent behavior**
<span style="color:#27ae60;">Child("ahmed")</span> triggers <span style="color:#3498db;">Parent.__init__(self, name)</span> (inherited). <span style="color:#9b59b6;">c.display()</span> calls the parent’s method — child has access to parent’s attributes and methods.

```python
c = Child("ahmed")
print(isinstance(c, Parent))
c.display() # call parent
```

---
### ⚠️ **Child overrides `__init__` without calling parent**
<span style="color:#3498db;">Employee</span> defines its own <span style="color:#9b59b6;">__init__(self, email)</span> and never calls <span style="color:#27ae60;">Person.__init__</span>. So <span style="color:#e74c3c;">Person</span> is not initialized. <span style="color:#e67e22;">Employee()</span> needs <span style="color:#1abc9c;">email</span> (TypeError if no args).

```python
# check this scenario ??
class Person:
    def __init__(self, name):
        self.name = name
        print("Person is created")

    def display(self):
        print(f"Person name is {self.name}")

class Employee(Person):
    def __init__(self, email):
        self.email = email
        print("Employee is created")

emp = Employee()
```

---
### ❌ **Bug: parent not initialized**
<span style="color:#e74c3c;">Employee.__init__</span> only sets <span style="color:#3498db;">email</span> and never calls <span style="color:#27ae60;">Person.__init__</span>, so <span style="color:#c0392b;">self.name</span> is never set. <span style="color:#e67e22;">emp.display()</span> then raises <span style="color:#c0392b;">AttributeError: 'Employee' object has no attribute 'name'</span>.

```python
emp = Employee("ahmed@gmail.com")

emp.display()
```

---
### ✅ **Calling parent with `super().__init__(name)`**
<span style="color:#27ae60;">Employee.__init__</span> takes <span style="color:#3498db;">name, email</span> and calls <span style="color:#9b59b6;">super().__init__(name)</span> so <span style="color:#e74c3c;">Person</span> sets <span style="color:#1abc9c;">self.name</span>. Then <span style="color:#e67e22;">emp.display()</span> works — parent and child are both initialized.

```python
# check this scenario ??
class Person:
    def __init__(self, name):
        self.name = name
        print("Person is created")

    def display(self):
        print(f"Person name is {self.name}")

class Employee(Person):
    def __init__(self,name,  email):
        super().__init__(name)
        self.email = email
        print("Employee is created")

emp = Employee("ahmed", "ahmed@gmail.com")
emp.display()
```

<h1 style='color:yellow;'> Multiple Inheritance </h1>

---
### 🧬 **Multiple inheritance — one child, two parents**
<span style="color:#27ae60;">class C(A, B):</span> means <span style="color:#e74c3c;">C</span> inherits from both <span style="color:#3498db;">A</span> and <span style="color:#e67e22;">B</span>. <span style="color:#9b59b6;">isinstance(cc, A)</span> and <span style="color:#9b59b6;">isinstance(cc, B)</span> are both <span style="color:#1abc9c;">True</span>.

```python
class A:
    pass

class B:
    pass


class C(A, B):
    pass

cc = C()
print(isinstance(cc, A))
print(isinstance(cc, B))
```

---
### 📌 **Only A has `__init__`**
<span style="color:#3498db;">A</span> has <span style="color:#9b59b6;">__init__</span>, <span style="color:#e67e22;">B</span> does not. <span style="color:#27ae60;">C(A, B)</span> inherits from both; when <span style="color:#e74c3c;">cc = C()</span>, A’s <span style="color:#1abc9c;">__init__</span> runs. Output: <span style="color:#95a5a6;">"A is created"</span>.

```python
class A:
    def __init__(self):
        print("A is created")

class B:
    pass


class C(A, B):
    pass

cc = C()
```

---
### 📌 **Only B has `__init__`**
<span style="color:#e67e22;">B</span> defines <span style="color:#9b59b6;">__init__</span>, <span style="color:#3498db;">A</span> does not. Creating <span style="color:#27ae60;">C()</span> uses B’s constructor (first in MRO that has one). Output: <span style="color:#1abc9c;">"B is created"</span>.

```python
class A:
    pass

class B:
    def __init__(self):
        print("B is created")


class C(A, B):
    pass

cc = C()
```

---
### 🔀 **Both parents have `__init__` — MRO decides**
<span style="color:#3498db;">C(A, B)</span> → Method Resolution Order is <span style="color:#27ae60;">C → A → B</span>. So <span style="color:#e74c3c;">A.__init__</span> runs first (A comes before B in the base list). Only <span style="color:#95a5a6;">"A is created"</span>.

```python
class A:
    def __init__(self):
        print("A is created")

class B:
    def __init__(self):
        print("B is created")


class C(A, B):
    pass

cc = C()
```

---
### 🏆 **Child defines its own `__init__`**
<span style="color:#27ae60;">C</span> has its own <span style="color:#9b59b6;">__init__</span>, so only <span style="color:#e74c3c;">"C is created"</span> prints. Parent <span style="color:#3498db;">A</span> and <span style="color:#e67e22;">B</span> initializers are **not** called unless we use <span style="color:#1abc9c;">super()</span>.

```python
class A:
    def __init__(self):
        print("A is created")

class B:
    def __init__(self):
        print("B is created")


class C(A, B):
    def __init__(self):
        print("C is created")

cc = C()
```

---
### 📞 **`super()` in multiple inheritance**
<span style="color:#27ae60;">C</span> calls <span style="color:#9b59b6;">super().__init__()</span> → that goes to <span style="color:#3498db;">A</span> (first base in MRO), not B. So only <span style="color:#e74c3c;">A</span> and <span style="color:#27ae60;">C</span> run. To call **both** parents, you’d need to call them explicitly or design the hierarchy so one chain runs.

```python
class A:
    def __init__(self):
        print("A is created")

class B:
    def __init__(self):
        print("B is created")


class C(A, B):
    def __init__(self):
        super().__init__()
        print("C is created")

cc = C()
```

<h1> Check this </h1>

---
### ⚠️ **Diamond: B does not call `super()`**
<span style="color:#3498db;">A</span> calls <span style="color:#9b59b6;">super()</span> → <span style="color:#e67e22;">B</span>; <span style="color:#e67e22;">B</span> does **not** call <span style="color:#9b59b6;">super()</span>, so <span style="color:#e74c3c;">Base</span> is never run. Output: <span style="color:#95a5a6;">B → A → C</span> (Base missing). Fix: always call <span style="color:#27ae60;">super().__init__()</span> in subclasses.

```python
class Base:
    def __init__(self):
        print("Base is created")
class A(Base):
    def __init__(self):
        super().__init__()
        print("A is created")

class B(Base):
    def __init__(self):
        print("B is created")


class C(A, B):
    def __init__(self):
        super().__init__()
        print("C is created")

cc = C()
```

---
### ✅ **Correct chain: both A and B call `super()`**
When <span style="color:#27ae60;">A</span> and <span style="color:#3498db;">B</span> both use <span style="color:#9b59b6;">super().__init__()</span>, the chain reaches <span style="color:#e74c3c;">Base</span>. Order: <span style="color:#1abc9c;">Base → B → A → C</span>. The MRO (Method Resolution Order) determines who is “next.”

```python
class Base:
    def __init__(self):
        print("Base is created")
class A(Base):
    def __init__(self):
        super().__init__()
        print("A is created")

class B(Base):
    def __init__(self):
        super().__init__()
        print("B is created")


class C(A, B):
    def __init__(self):
        super().__init__()
        print("C is created")

cc = C()
```

# overriding

---
### 🔄 **Method overriding**
<span style="color:#9b59b6;">Employee.display()</span> **overrides** <span style="color:#3498db;">Person.display()</span> but still calls <span style="color:#27ae60;">super().display()</span> to reuse the parent logic, then adds extra output. Override = same method name, custom behavior in the child.

```python
emp.display()


# check this scenario ??
class Person:
    def __init__(self, name):
        self.name = name
        print("Person is created")

    def display(self):
        print(f"Person name is {self.name}")


class Employee(Person):
    def __init__(self, name, email):
        super().__init__(name)
        self.email = email
        print("Employee is created")

    def display(self):
        print(f"-------------------------------------")
        super().display()
        print("---- this employee is displayed----")
        print(f"-------------------------------------")



emp = Employee("ahmed", "ahmed@gmail.com")
emp.display()
```

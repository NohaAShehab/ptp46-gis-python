- read data

```python
try:
    """ opening file with read mode open it for reading """
    file_object = open('users.txt','r')
    print(file_object)
    file_data= file_object.read() # read file content into one string
    print(file_data, type(file_data))
    file_object.close()


except Exception as e:
    print(e)
```

- read file line by line ?

```python
try:
    """ opening file with read mode open it for reading """
    file_object = open('users.txt','r')
    print(file_object)
    file_data= file_object.readlines() # read file content into
    print(file_data, type(file_data))
    file_object.close()


except Exception as e:
    print(e)
```

```python
try:
    """ opening file with read mode open it for reading """
    file_object = open('users.txt','r')
    print(file_object)
    file_data= file_object.readlines() # read file content into
    print(file_data, type(file_data))
    file_object.seek(0) # reset file pointer to the beginning of the file === byte==0
    data = file_object.read()
    print(data, type(data))
    file_object.close()


except Exception as e:
    print(e)
```

```python
try:
    """ opening file with read mode open it for reading """
    with open('users.txt','r') as file:
        data = file.read()
        print(data, type(data))
        print(file)


except Exception as e:
    print(e)
```

<h1 style='color:yellow; text-align:center'>Save data to a file  </h1>

```python
try:
    """ opening file with write mode open it for writing
    why you try to open file with write mode if file doesn't exist python will try to create it ...
    """
    file_object = open('students.txt','w')
    file_object.write('Hello from Python')
    file_object.close()

except Exception as e:
    print(e)
```

```python
try:
    """ opening file with write mode open it for writing
    why you try to open file with write mode if file doesn't exist python will try to create it ...
    if file exists --> python will open file for writing starting from  the beginning of the file ... remove old content.
    """
    file_object = open('students.txt','w')
    file_object.close()

except Exception as e:
    print(e)
```

```python
try:
    """
    when you try to open file with write mode if file doesn't exist python will try to create it ...
    if file exists --> python will open file for writing starting from  the beginning of the file ... remove old content.
    """
    file_object = open('students.txt','w')
    file_object.writelines(["hello", "from", 'python',"iti"])
    file_object.close()

except Exception as e:
    print(e)
```

```python
try:
    """ 
    when you try to open file with write mode if file doesn't exist python will try to create it ...
    if file exists --> python will open file for writing starting from  the beginning of the file ... remove old content.
    """
    file_object = open('students.txt','w')
    file_object.writelines(["hello\n", "from\n", 'python\n',"iti\n"])
    file_object.close()

except Exception as e:
    print(e)
```

<h1 style='color:orange; text-align:center'>Append data to a file  </h1>

```python
try:
    """
    when you try to open file with append mode if file doesn't exist python will try to create it ...
    if file exists --> python will open file for writing starting from  the end of the file ... keep old content.
    """
    file_object = open('students.txt','a')
    file_object.writelines("Hello from Appned \n")
    file_object.close()

except Exception as e:
    print(e)
```

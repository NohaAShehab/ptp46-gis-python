```python
# list is mutable datatype ??

# append element to the list ?
myl.append("new item")
```

```python
print(myl)
```

```python
myl.insert(4, "inserted item")
print(myl)
```

```python
# update value at existing index ?
myl[3]="update value"
print(myl)
```

```python
# extend a list ??
courses = ["c", "ml", "dl", "interview skills"]

for item in courses:
    myl.append(item)

print(myl)
```

```python
myl.extend(courses)
print(myl)
```

```python
# remove element from list ?
# pop
print(myl)
popped_element = myl.pop() # remove last element from list
print(popped_element)
print(myl)
```

```python
# pop element at index
popped_element2 = myl.pop(5)
print(popped_element2)
print(myl)
```

```python
# if you want to remove element without knowing its index --> remove ?
print(myl)

myl.remove('iti') # remove first occurrence of the element

print(myl)
```

```python
print(myl.index("iti")) # get index of the first occurrence of the element
```

```python
# iterable

for item in myl:
    print(f"item = {item}")
```

```python
for index, item in enumerate(myl):
    print(f"index = {index} , item = {item}")
```

```python
# concat 2 lists

l1 = ["Hussien", "Abdulrahman "*3 , "Ahmed" ]
print(l1 )
```

```python
l2 = ["abdelrahman"] *3
print(l2 )
```

```python
l3 = l1 + l2
print(l3)
```

```python
# sorting list
myl.sort() # sorting of list elements --> requires that all element in the list must have the same type
```

```python
names = ["Ahmed", "hussien", 'Essam', 'Abdelrahman',
         'sobhy', 'hager', 'Mai']

names.sort()  # sort at the same variable , ascending
print(names)
```

```python
names = ["Ahmed", "hussien", 'Essam', 'Abdelrahman',
         'sobhy', 'hager', 'Mai']

names.sort(reverse=True)

print(names)
```

```python
# you can reverse a list ?
print(myl)
myl.reverse() # in the same variable
print(myl)
```

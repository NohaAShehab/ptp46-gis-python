```python
# mail contain only one @, contain domain .com, .eg  at least one . ?
# noha.a.shehab@gmail.com
# noha-dd@yahoo.com  # no only _
# google

# test@fff-ddd-ff.com
#
import re
pattern = r'[A-Za-z0-9]+@[\w]+\.[\w]+'
email = 'noha'
valid_email = re.match(pattern, email)
print(valid_email)
```

```python
user_email = 'noha@iti.gov.eg'
valid_email = re.match(pattern, user_email)
print(valid_email)
```

```python
user_email2 = 'noha@iti.gov.eg@iti.gov.eg'
valid_email2 = re.match(pattern, user_email2)
print(valid_email2)
```

```python
user_email3 = 'noha@iti.gov.eg@iti.gov.eg'
valid_email3 = re.fullmatch(pattern, user_email3)
print(valid_email3)
```

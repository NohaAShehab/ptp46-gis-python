# Add validation to the function before performing the operation ?

```python
def sum_num(num1, num2):
    if isinstance(num1, int) and isinstance(num2, int):
        res = num1 + num2
        return res
    return "Not valid parameters"

rr = sum_num(3,5)
print(rr)
```

```python
r5 = sum_num("iti", "gis")
print(r5)
```

```python
def summ(num1 = 0, num2 = 0):
    print(f"num1: {num1}, num2: {num2}")
    res = num1 + num2
    return res

a1 = summ(34,5)
print(a1)
#############
a2 = summ()
print(a2)

#############
a3 = summ(30)
print(a3)
```

# define function accept unknown number of key:word arguments

```python
def introduce_your_self(**info):
    print(info)  # save data to a dict.

introduce_your_self()
introduce_your_self(name='ahmed')
introduce_your_self(fname='ahmed', age=22)
```

```python
def introduce_your_self(**info):
    print(info)  # save data to a dict.
    for k, v in info.items():
        print(f"{k}: {v}")

    if "name" in info.keys():
        print(info["name"])


introduce_your_self()
introduce_your_self(name='ahmed')
introduce_your_self(fname='ahmed', age=22)
```
